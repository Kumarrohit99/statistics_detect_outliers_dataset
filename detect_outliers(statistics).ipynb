```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
dataset = [
    11, 10, 12, 14, 12, 15, 14, 13, 15, 102,
    12, 14, 17, 19, 107, 10, 13, 12, 14, 12,
    108, 12, 11, 14, 13, 15, 10, 15, 12, 10,
    14, 13, 15, 10
]
```

```python
dataset.sort()
print(dataset)
```

```python
Q1 = np.percentile(dataset,25)
print(Q1)
```

```python
Q3 = np.percentile(dataset,75)
print(Q3)
```

```python
IQR = Q3-Q1
```

```python
lower_fence = Q1-1.5*IQR
print(lower_fence)
```

```python
higher_fence = Q3+1.5*IQR
print(higher_fence)
```

```python
outliers = []

for i in dataset:
    if(i<lower_fence) or (i>higher_fence):
        outliers.append(i)
print(outliers)
   
```

```python
import seaborn as sns
sns.boxplot(dataset)
```

```python
outliers = []
#z-score

def detect_outliers(data):
    threshold=3 ## third standard deviation
    mean = np.mean(data)
    std = np.std(data)

    for i in data:
        z_score = (i-mean)/std
        if np.abs(z_score)>threshold:
            outliers.append(i)
    return outliers
```

```python
detect_outliers(dataset)
print(outliers)
```
